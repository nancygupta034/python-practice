```python
#Level 1 (Basic NumPy Practice)
#Question 1 — Prediction + Error

import numpy as np

X = np.array([1,2,3,4])
y = np.array([5,7,9,11])

m = 2
b = 3

#y_pred calculate karo
y_pred = m*X + b
print(y_pred)

#error calculate karo
error = y - y_pred
print(error)

#MSE calculate karo.
mse = np.mean(error**2)
print(mse)
```

```python
#Question 2 — Loss Function
X = np.array([2,4,6])
y = np.array([10,14,18])

m = 2
b = 4

#1️⃣ predictions
y_pred = m*X + b

#2️⃣ error
error = y - y_pred

#3️⃣ squared error
sq_error = error**2

#4️⃣ MSE
mse = np.mean(sq_error)
print(mse)
```

```python
#Level 2 (Gradient Practice)
#Question 3 — Gradient Calculation

X = np.array([1,2,3])
y = np.array([2,4,6])
m = 0
b = 0
n = len(X)

y_pred = m*X + b
error = y - y_pred

#Compute:
dm = (-2/n) * np.sum(X * error)
db = (-2/n) * np.sum(error)

print("dm => ", dm)
print("db => ",  db)
```

```python
#Question 4 — One Gradient Descent Step
X = np.array([1,2,3])
y = np.array([4,6,8])
m = 0
b = 0
lr = 0.01
y_pred = m*X + b
error = y - y_pred

dm = (-2/n)*np.sum(X*error)
db = (-2/n)*np.sum(error)

m = m - lr * dm
b = b - lr * db
```

```python
#Level 3 (Full Training)
#Question 5 — Train Linear Regression (NumPy)

X = np.array([1,3,5,7])
y = np.array([10,14,18,22])
m = 0
b = 0
lr = 0.01
n = len(X)

for i in range(2000):
    y_pred = m*X + b
    error = y - y_pred
    loss = np.mean(error**2)

    dm = (-2/n)*np.sum(X*error)
    db = (-2/n)*np.sum(error)

    m = m-lr*dm
    b = b-lr*db


print("Final m => ", abs(m))
print("Final b => ", abs(b))
```
