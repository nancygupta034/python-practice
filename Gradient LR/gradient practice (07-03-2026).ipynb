```python
#Question 1 (Very Easy)
import numpy as np

X = np.array([2,4,6,8])
y = np.array([50,70,90,110])
#Model parameters:
m = 10
b = 30

#Predictions calculate karo
y_pred = (m * X) + b
print("y_pred => ", y_pred)

#Error nikaalo
error = y - y_pred
print("error => ", error)

#MSE (Mean Squared Error) calculate karo.
loss = np.mean(error**2)
print("loss => ", loss)
```

```python
#Question 2 (Prediction + Loss)
X = np.array([1,3,5,7])
y = np.array([40,60,80,100])

#Parameters:
m = 8
b = 25

print("y => ", y)
#y_pred find karo
y_pred = (m * X) + b
print("y_pred => ", y_pred)

#error find karo
error = y - y_pred
print("error => ", error)

#Squared error nikaalo
sq_error = error ** 2
print("sq_error => ", sq_error)

#MSE find karo using NumPy
mse = np.mean(sq_error)
print("mse => ", mse)
```

```python
#Question 3 (Gradient Calculation)

X = np.array([1,2,3])
y = np.array([2,4,6])

#Initial parameters:
m = 0
b = 0
n = len(X)

#NumPy use karke calculate karo:
#y_pred
y_pred = (m * X) + b

#error
error = y - y_pred

#gradient of m   => dm=(−2/n)∗Σ(X∗error)
dm = (-2/n) * np.sum( X * error)
print("dm => ", dm)


#gradient of b => db=(−2/n)∗Σ(error)
db = (-2/n) * np.sum(error)
print("db => ", db)


```

```python
#Question 4 (One Gradient Descent Step)

#Dataset
X = np.array([1,2,3])
y = np.array([3,5,7])

#Initial:
m = 0
b = 0
lr = 0.01
n = len(X)

#Predict y_pred
y_pred = (m * X) + b
print(y_pred)

#Find error
error = y - y_pred
print("error => ", error)

#Calculate gradients (dm, db)
dm = (-2/n) * np.sum(X * error)
print("dm => ", dm)
db = (-2/n) * np.sum(error)
print("db => ", db)

#Update parameters
m = m - lr * dm
b = b - lr * db
```

```python
#Question 5 (Mini Linear Regression)

#Dataset:
X = np.array([1,2,4,5])
y = np.array([35,45,65,75])

#Tasks => NumPy use karke:

#m = 0, b = 0 se start karo
m = 0
b = 0
n = len(X)
lr = 0.01

#1000 iterations gradient descent run karo
for i in range(1000):

    y_pred = m*X + b
    error = y - y_pred
    loss = np.mean(error**2)

    dm = (-2/n) * np.sum(X * error)
    db = (-2/n) * np.sum(error)

    m = m - lr*dm
    b = b - lr*db

    if i % 100 == 0:
        print("Iteration:", i, "Loss:", loss)

print("Final m:", m)
print("Final b:", b)


```
