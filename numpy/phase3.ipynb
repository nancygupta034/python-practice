## Advanced operation with business example

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 280000],
    [2, 120000, 140000, 160000, 190000],
    [3, 200000, 230000, 260000, 300000],
    [4, 180000, 210000, 240000, 270000],
    [5, 160000, 185000, 205000, 230000],

])

print("=====Zomato sales analysis======")
print("Shape is : ", sales_data.shape)
print("Sample data of first 3 restaurants: ", sales_data[:3])
```

```python
# total sales per year
print(np.sum(sales_data, axis=0))
yearly_total = np.sum(sales_data[:, 1:],axis=0)
print("YEalry total second method: ", yearly_total)

# minimum sales per restaurant
min_sales = np.min(sales_data[:, 1:], axis=1)
print("minimum sales per restaurant: ", min_sales)

# maximum sales per year
max_sales = np.max(sales_data[:, 1:], axis=0)
print("maximum sales per year", max_sales)

# average sales per restaurant
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print("average sales per restaurant: ", avg_sales)
```

```python
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(cumsum)

plt.figure(figsize=(8,6))
plt.title("Average cumulative sales accross the restaurants")
plt.plot(np.mean(cumsum, axis=0))
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
