```python
import pandas as pd

data = {
    "name": ["Aman", "Riya", "Karan", "Neha", "Aman"],
    "age": [23, 25, 22, 24, 23],
    "city": ["Delhi", "Mumbai", "Delhi", "Pune", "Delhi"],
    "salary": [40000, 50000, 35000, 45000, 40000]
}

df = pd.DataFrame(data)
```

```python
#Show data types
df.dtypes
```

```python
#Select only name column
df["name"]
```

```python
#Select name and salary columns
df[["name", "salary"]]
```

```python
#Get ages greater than 23
df[df["age"] > 23]
```

```python
#Get rows where city is "Delhi"
df[df["city"] == "Delhi"]
```

```python
#Get rows where salary > 40000
df[df["salary"] > 40000]
```

```python
#Get rows where city is Delhi AND age < 24
df[(df["city"] == "Delhi") & (df["age"] < 24)]
```

```python
#Sort by salary ascending
df.sort_values("salary")
```

```python
#Sort by age descending
df.sort_values("age", ascending=False)
```

```python
#Find average salary
df["salary"].mean()
```

```python
#Find maximum age
df["age"].max()
```

```python
#Count how many people are from Delhi
(df["city"] == "Delhi").sum()
```

```python
#Get average salary per city
df.groupby("city")["salary"].mean()
```

```python
#Count number of people per city
df.groupby("city")["name"].count()
```

```python
#Check if any duplicate rows exist
df.duplicated().any()
```

```python
#Remove duplicates
df.drop_duplicates(inplace=True)
df
```

```python
#Create new column tax = salary * 0.1
df["tax"] = df["salary"] * 0.1
df
```

```python
#Create new column age_after_5_years
df["age_after_5_years"] = df["age"] + 5
df
```

```python
#Rename salary to monthly_salary
df.rename(columns={"salary":"monthly_salary"}, inplace=True)
df
```
