```python
import pandas as pd

data = {
    "user_id": [101, 102, 103, 101, 104, 102, 105],
    "prompt": [
        "Generate Python code",
        "Translate this text",
        None,
        "Summarize article",
        "Generate Python code",
        "Fix bug in code",
        "Translate this text"
    ],
    "category": ["coding", "translation", "writing", "writing", "coding", "coding", "translation"],
    "response_time": [2.5, 3.2, 4.1, None, 2.8, 5.5, 3.0],
    "tokens_used": [250, 180, 300, 220, 260, 400, 190]
}

df = pd.DataFrame(data)
print(df)
```

```python
#First 3 rows
first_3_rows = df.head(3)
print(first_3_rows)
#Last 2 rows
last_2_rows = df.tail(2)
print(last_2_rows)
#Column names
columns_name = df.columns
print(columns_name)
#Data type
print(df.dtypes)
```

```python
#Get rows where category == "coding"
df[df["category"] == "coding"]
```

```python
#Get rows where response_time > 3
df[df["response_time"] > 3]
```

```python
#Get rows where tokens_used >= 250
df[df["tokens_used"] >= 250]
```

```python
print(df)
```

```python
#Count missing values per column
df.isnull().sum()
```

```python
#Replace missing response_time with mean
df["response_time"].fillna(df["response_time"].mean())
```

```python
#Drop rows where prompt is missing
df.dropna(subset="prompt")
```

```python
#Convert all prompts to lowercase
df["prompt"] = df["prompt"].str.lower()
df
```

```python
#Create new column prompt_length (character count)
df["prompt_length"] = df["prompt"].str.len()
df
```

```python
#Create new column word_count
df["word_count"] = df["prompt"].str.split().str.len()
df
```

```python
#Total tokens used per user
df.groupby("user_id")["tokens_used"].sum()
```

```python
#Average response time per category
df.groupby("category")["response_time"].mean()
```

```python
#Count prompts per category
df.groupby("category")["prompt"].count()
```

```python
#Sort by tokens_used descending
df.sort_values("tokens_used", ascending=False)

```

```python
#Get top 2 highest token usage rows
df.sort_values("tokens_used", ascending=False).head(2)
```
